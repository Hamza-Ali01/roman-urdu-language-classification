```python
import pandas as pd

# Load dataset
df = pd.read_csv("../dataset.csv")

print("Dataset loaded successfully!")
print("Total word entries:", len(df))
print("Total sentences:", df["Sentence_ID"].nunique())

print("\nColumns:")
print(df.columns.tolist())

print("\nLabel distribution:")
print(df["Label"].value_counts())
```

```python
print("First 10 rows:")
display(df.head(10))

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("=== Dataset Statistics ===")

print("\nTotal word entries:", len(df))
print("Total sentences:", df["Sentence_ID"].nunique())

print("\nLabel Distribution:")
print(df["Label"].value_counts())

print("\nLabel Percentage:")
print(df["Label"].value_counts(normalize=True).mul(100).round(2))

print("\nSample Sentences:")
for sentence_id in df["Sentence_ID"].unique()[:5]:
    sentence = df[df["Sentence_ID"] == sentence_id]["Sentence"].iloc[0]
    print(f"{sentence_id}: {sentence}")
```
